```python
# Vehicle Service Management System

from datetime import datetime

class Owner:
    def __init__(self, name, phone):
        self.__name = name
        self.__phone = phone

    def __str__(self):
        return f"Owner: {self.__name}, Phone: {self.__phone}"

    def getName(self):
        return self.__name

owner1 = Owner("Allison Huber", "636-634-6599")

class Vehicle:
    def __init__(self, model, year, registration_number, owner):
        self.__model = model
        self.__year = year
        self.__registration_number = registration_number
        self.__owner = owner

    def __str__(self):
        return (f"Vehicle: {self.__model} " 
                f"({self.__year}), " 
                f"Reg#: {self.__registration_number}, " 
                f"Owner: {self.__owner.getName()}")

    def getModel(self):
        return self.__model

vehicle1 = Vehicle ("Subaru Forester", 2020, "DV662041SR", owner1)

class Mechanic:
    def __init__(self, name, specialization):
        self.__name = name
        self.__specialization = specialization

    def __str__(self):
        return (f"Mechanic: {self.__name}, " 
                f"Specialization: {self.__specialization}")

    def getName(self):
        return self.__name

mechanic1 = Mechanic("Robert", "Engine Repair")

class ServiceAppointment:
    def __init__(self, vehicle, mechanic, service_date):
        self.__vehicle = vehicle
        self.__mechanic = mechanic
        self.__service_date = service_date
        self.__status = "Pending"
        self.__completion_time = None

    def mark_completed(self):
        self.__status = "Completed"
        self.__completion_time = datetime.now()

    def get_duration(self):
        if self.__completion_time is None:
            return "Service not completed yet"
        return self.__completion_time - self.__service_date

    def __str__(self):
        return (f"Appointment for {self.__vehicle.getModel()} " 
                f"on {self.__service_date.strftime('%Y-%m-%d %H:%M')} " 
                f"with {self.__mechanic.getName()} | " 
                f"Status: {self.__status}")

    def getStatus(self):
        return self.__status
        
appointment1 = ServiceAppointment(vehicle1, mechanic1, datetime(2026, 6, 27, 10, 30))

class ServiceCenter:
    def __init__(self):
        self.__owners = []
        self.__vehicles = []
        self.__mechanics = []
        self.__appointments = []

    def register_owner(self, owner):
        self.__owners.append(owner)
        print("Registered Owner:")
        print(owner)

    def add_vehicle(self, vehicle):
        self.__vehicles.append(vehicle)
        print("Added Vehicle:")
        print(vehicle)

    def hire_mechanic(self, mechanic):
        self.__mechanics.append(mechanic)
        print("Hired Mechanic:")
        print(mechanic)

    def schedule_service(self, appointment):
        self.__appointments.append(appointment)
        print("Scheduled Appointment:")
        print(appointment)

    def list_active_appointments(self):
        print("Active Appointments:")
        for appointment in self.__appointments:
            if appointment.getStatus() == "Pending":
                print(appointment)
                
    def complete_service(self, appointment):
        print("Completing service...")
        appointment.mark_completed()

        print("Updated Appointment:")
        print(appointment)

        print("Service Duration:", appointment.get_duration())

service_center = ServiceCenter()

service_center.register_owner(owner1)
service_center.add_vehicle(vehicle1)
service_center.hire_mechanic(mechanic1)
service_center.schedule_service(appointment1)
service_center.list_active_appointments()
service_center.complete_service(appointment1)
```
